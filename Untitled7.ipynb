```python

# Penyelesaian SPL dengan Metode Eliminasi Gauss

def eliminasi_gauss(A, n):
    EPS = 1e-12

    # Tahap 1: Eliminasi maju
    for k in range(n - 1):

        # Mencari pivot terbesar (partial pivoting)
        p = max(range(k, n), key=lambda i: abs(A[i][k]))

        if abs(A[p][k]) < EPS:
            print("SPL tidak memiliki solusi tunggal.")
            return None

        # Menukar baris jika diperlukan
        if p != k:
            A[k], A[p] = A[p], A[k]

        # Menghilangkan elemen di bawah pivot
        for i in range(k + 1, n):
            faktor = A[i][k] / A[k][k]

            for j in range(k, n + 1):
                A[i][j] -= faktor * A[k][j]

    if abs(A[n-1][n-1]) < EPS:
        print("SPL tidak memiliki solusi tunggal.")
        return None

    print("\nMatriks setelah eliminasi:")
    for baris in A:
        print([round(nilai, 4) for nilai in baris])

    # Tahap 2: Substitusi balik
    x = [0.0] * n

    for i in range(n - 1, -1, -1):
        jumlah = A[i][n]

        for j in range(i + 1, n):
            jumlah -= A[i][j] * x[j]

        x[i] = jumlah / A[i][i]

    return x


# Program Utama
print("=== METODE ELIMINASI GAUSS ===")

n = int(input("Masukkan jumlah variabel: "))

if n < 1:
    print("Jumlah variabel minimal 1.")
else:
    A = []

    print("\nMasukkan koefisien dan konstanta:")
    print("Pisahkan setiap angka dengan spasi.")

    for i in range(n):
        while True:
            baris = list(map(float, input(
                f"Persamaan ke-{i+1}: "
            ).split()))

            if len(baris) == n + 1:
                A.append(baris)
                break
            else:
                print(f"Harus memasukkan {n+1} angka!")

    hasil = eliminasi_gauss(A, n)

    if hasil is not None:
        print("\n=== HASIL PENYELESAIAN ===")

        for i in range(n):
            print(f"x{i+1} = {hasil[i]:.4f}")
```
